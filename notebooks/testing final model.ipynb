```python
import pandas as pd

data = pd.read_csv("data.csv")

print("Dataset loaded successfully.")
print(data.shape)
```

```python
data=data.drop_duplicates()
```

```python
data["Time"]=data["Time"]/3600
```

```python
from sklearn.model_selection import train_test_split
```

```python
X=data.drop(columns=['Class'])
y=data['Class']
```

```python
X_train,X_test,y_train,y_test= train_test_split(X,y,test_size=0.2,stratify=data['Class'],random_state=42)
```

```python
sample = X_test.iloc[[0]].copy()
actual = y_test.iloc[0]

print("Actual class:", actual)
display(sample)
```

```python
import joblib
```

```python
loaded_model = joblib.load("models/fraud_detection_model.pkl")
loaded_amount_transformer = joblib.load("models/amount_transformer.pkl")
loaded_scaler = joblib.load("models/scaler.pkl")

print("Models loaded successfully.")
```

```python
sample["Amount"]=loaded_amount_transformer.transform(sample[["Amount"]])
```

```python
sample[["Time","Amount"]]=loaded_scaler.transform(sample[["Time","Amount"]])
```

```python
prediction = loaded_model.predict(sample)[0]
probability = loaded_model.predict_proba(sample)[0, 1]
```

```python
print("Actual class:", actual)
print("Predicted class:", prediction)
print("Fraud probability:", probability)
```
